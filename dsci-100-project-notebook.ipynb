```r
# Select all numeric variables
numeric_data <- march_madness_tidy |>
    select(year, seed, reg_wins_pct, tourney_wins)

# Calculate summary statistics for each variable
march_madness_summary <- numeric_data |>
    map_df(
        ~ tibble(
            mean = mean(.x, na.rm = TRUE),
            min = min(.x, na.rm = TRUE),
            max = max(.x, na.rm = TRUE)
        ),
        .id = "variable"
    )

# Display the summary table
march_madness_summary
```

```r
# Plot the distribution of tournament wins
tourney_wins_plot <- march_madness_tidy |>
    ggplot(aes(x = tourney_wins)) +
    geom_histogram(binwidth = 1) +
    labs(
        title = "Distribution of Tournament Wins",
        x = "Number of Tournament Wins",
        y = "Number of Tournament Appearances"
    )

tourney_wins_plot
```

```r
# Convert seed into a categorical variable for plotting
# Add tournament seed as a categorical variable
march_madness_tidy <- march_madness_tidy |>
    mutate(seed_category = as_factor(seed))

# Calculate the proportion of each tournament seed
seed_proportions <- march_madness_tidy |>
    group_by(seed_category) |>
    summarize(count = n()) |>
    mutate(proportion = count / sum(count))

# Plot the relative proportions
seed_plot <- seed_proportions |>
    ggplot(aes(x = seed_category, y = proportion)) +
    geom_col() +
    labs(
        title = "Proportion of Tournament Appearances by Seed",
        x = "Tournament Seed",
        y = "Proportion of Tournament Appearances"
    )

# Display the plot
seed_plot
```

This visualization shows the relationship between tournament seed and average tournament wins. It helps us examine whether teams with lower-numbered seeds tend to win more games. This is relevant to our research question because tournament seed is one of the predictors we want to use to predict tournament wins.

```r
# Calculate average tournament wins for each seed
wins_by_seed <- march_madness_tidy |>
    group_by(seed) |>
    summarize(mean_tourney_wins = mean(tourney_wins, na.rm = TRUE))

# Plot the relationship between seed and tournament wins
seed_wins_plot <- wins_by_seed |>
    ggplot(aes(x = seed, y = mean_tourney_wins)) +
    geom_point() +
    geom_line() +
    labs(
        title = "Average Tournament Wins by Tournament Seed",
        x = "Tournament Seed",
        y = "Average Number of Tournament Wins"
    )

seed_wins_plot
```

This visualization shows a positive relationship between regular-season winning percentage and tournament wins. Teams with more tournament wins generally have higher average regular-season winning percentages. This suggests that regular-season winning percentage could be a useful predictor of tournament success, which is directly related to our research question.

```r
# Calculate average regular-season winning percentage
# for each number of tournament wins
win_pct_summary <- march_madness_tidy |>
    group_by(tourney_wins) |>
    summarize(
        mean_reg_wins_pct = mean(reg_wins_pct, na.rm = TRUE)
    )

# Plot the relationship
win_pct_plot <- win_pct_summary |>
    ggplot(aes(x = as_factor(tourney_wins),
               y = mean_reg_wins_pct)) +
    geom_col() +
    labs(
        title = "Average Regular-Season Winning Percentage by Tournament Wins",
        x = "Number of Tournament Wins",
        y = "Average Regular-Season Winning Percentage (%)"
    )

win_pct_plot
```
